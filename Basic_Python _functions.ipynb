Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.

```python
def analyze_text(text):
    vowels = 0
    digits = 0
    spaces = 0
    for char in text:
        if char.lower() in "aeiou":
            vowels += 1
        if char.isdigit():
            digits += 1
        if char == " ":
            spaces += 1
    result = {
        "characters": len(text),
        "words": len(text.split()),
        "vowels": vowels,
        "digits": digits,
        "spaces": spaces
    }
    return result

text = input("Enter text: ")
text1 = analyze_text(text)
print("Total characters:", text1["characters"], "Total words:", text1["words"], "Total vowels:", text1["vowels"], "Total digits:", text1["digits"], "Total spaces:", text1["spaces"],end="")

```

```python
def generate_fibonacci(terms):
    if terms<=0:
        return []
    fibonacci=[0,1]
    for i in range(2,terms):
        next_number=fibonacci[i-1]+fibonacci[i-2]
        fibonacci.append(next_number)
    return fibonacci
terms=int(input("Enter number of terms:"))
result=generate_fibonacci(terms)
print(result)
```

```python
from functools import reduce

company_name="Edukron"

employees=[
    {"id":101,"name":"Ravi","department":"IT","salary":60000,"performance_score":88,"experience":5},
    {"id":102,"name":"Anitha","department":"HR","salary":45000,"performance_score":72,"experience":4},
    {"id":103,"name":"Kiran","department":"IT","salary":75000,"performance_score":94,"experience":7},
    {"id":104,"name":"Meena","department":"Finance","salary":55000,"performance_score":81,"experience":6},
    {"id":105,"name":"Arjun","department":"IT","salary":40000,"performance_score":65,"experience":2}
]

# Part 1: Display all employees
def display_employees(data):
    if len(data)==0:
        print("No employees found")
        return

    for employee in data:
        print(employee)

# Part 2: Get employee names
def get_names(data):
    return list(map(lambda employee:employee["name"],data))

# Part 3: High performers
def get_high_performers(data):
    return list(filter(lambda employee:employee["performance_score"]>=80,data))

# Part 4: Search by department
def search_department(data,department):
    return list(filter(lambda employee:employee["department"].lower()==department.lower(),data))

# Part 5: Calculate revised salaries
def calculate_revised_salaries(data):
    def calculate(employee):
        score=employee["performance_score"]
        salary=employee["salary"]

        if score>=90:
            increase=15
        elif score>=80:
            increase=10
        elif score>=70:
            increase=5
        else:
            increase=0

        new_salary=salary+(salary*increase/100)

        return {
            "name":employee["name"],
            "old_salary":salary,
            "increase":increase,
            "new_salary":new_salary
        }

    return list(map(calculate,data))

# Part 6: Calculate salary expenses
def salary_expense(data):
    if len(data)==0:
        return 0,0,0

    current_total=reduce(lambda a,b:a+b["salary"],data,0)

    revised=calculate_revised_salaries(data)

    revised_total=reduce(lambda a,b:a+b["new_salary"],revised,0)

    additional=revised_total-current_total

    return current_total,revised_total,additional

# Part 7: Highest paid employee
def highest_paid(data):
    if len(data)==0:
        return None

    return reduce(
        lambda a,b:a if a["salary"]>=b["salary"] else b,
        data
    )

# Part 8: Sorting employees
def sort_employees(data):
    print("\nSorted by name:")
    print(sorted(data,key=lambda employee:employee["name"]))

    print("\nSorted by salary:")
    print(sorted(data,key=lambda employee:employee["salary"]))

    print("\nSorted by performance:")
    print(sorted(data,key=lambda employee:employee["performance_score"]))

    print("\nSorted by experience:")
    print(sorted(data,key=lambda employee:employee["experience"]))

# Part 9: Closure based bonus calculator
def create_bonus_calculator(percentage):
    def calculate_bonus(salary):
        return salary*percentage/100

    return calculate_bonus

bonus5=create_bonus_calculator(5)
bonus10=create_bonus_calculator(10)
bonus15=create_bonus_calculator(15)

# Part 10 and 11: Department report and LEGB
def generate_department_report(data,department):
    # Enclosing scope variable
    department_name=department

    def employee_count():
        return len(data)

    def total_salary():
        return reduce(lambda a,b:a+b["salary"],data,0)

    def average_salary():
        if len(data)==0:
            return 0
        return total_salary()/len(data)

    def average_performance():
        if len(data)==0:
            return 0

        total=reduce(
            lambda a,b:a+b["performance_score"],
            data,
            0
        )

        return total/len(data)

    # Local scope variable
    report_title="Department Report"

    # LEGB:
    # L = Local: report_title is found in this function.
    # E = Enclosing: department_name belongs to the outer report function.
    # G = Global: company_name is found outside all functions.
    # B = Built-in: len(), print() and reduce-related built-ins are Python built-ins.

    return {
        "company":company_name,
        "title":report_title,
        "department":department_name,
        "employees":employee_count(),
        "total_salary":total_salary(),
        "average_salary":average_salary(),
        "average_performance":average_performance()
    }

# Part 12: Menu
def menu():
    while True:
        print("\n===== Employee Performance Analysis =====")
        print("1. Display all employees")
        print("2. Display employee names")
        print("3. Display high performers")
        print("4. Search by department")
        print("5. Calculate revised salaries")
        print("6. Calculate total salary expense")
        print("7. Find highest-paid employee")
        print("8. Sort employees")
        print("9. Generate department report")
        print("10. Exit")

        choice=input("Enter your choice:")

        if choice=="1":
            display_employees(employees)

        elif choice=="2":
            print("Employee names:",get_names(employees))

        elif choice=="3":
            result=get_high_performers(employees)
            display_employees(result)

        elif choice=="4":
            department=input("Enter department:")
            result=search_department(employees,department)

            if len(result)==0:
                print("Department not found")
            else:
                display_employees(result)

        elif choice=="5":
            result=calculate_revised_salaries(employees)

            for employee in result:
                print(employee)

        elif choice=="6":
            current,revised,additional=salary_expense(employees)

            print("Current total salary:",current)
            print("Revised total salary:",revised)
            print("Additional salary expense:",additional)

        elif choice=="7":
            employee=highest_paid(employees)

            if employee is None:
                print("No employees found")
            else:
                print("Highest-paid employee:",employee)

        elif choice=="8":
            sort_employees(employees)

        elif choice=="9":
            department=input("Enter department:")
            result=search_department(employees,department)

            if len(result)==0:
                print("Department not found")
            else:
                report=generate_department_report(result,department)

                for key,value in report.items():
                    print(key,":",value)

        elif choice=="10":
            print("Program ended")
            break

        else:
            print("Invalid choice. Please enter 1 to 10.")

menu()
```
