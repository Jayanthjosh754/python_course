50. Write a Python program that uses multiple built-in functions together to process a list of numbers.

```python
a=[10,30,40,50,60]
list_length = len(a)
list_sum = sum(a)
list_min = min(a)
list_max = max(a)
print(f"original list: {a}")
print(f"sum of list: {list_sum}")
print(f"min of list: {list_min}")
print(f"max of list: {list_max}")
print(f"length of list: {list_length}")
```
